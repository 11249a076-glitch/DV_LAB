```r
import matplotlib.pyplot as plt
subjects = ["Python", "Java", "C#", "AI"]
students = [40, 35, 25, 45]
plt.bar(subjects, students)
plt.title("Bar Chart")
plt.show()
```

```r
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 150, 130, 170]
plt.plot(months, sales)
plt.title("Line Chart")
plt.show()
```

```r
import matplotlib.pyplot as plt

labels = ["CSE", "IT", "ECE", "EEE"]
values = [120, 90, 80, 60]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.show()
```

```r
import matplotlib.pyplot as plt
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]
plt.scatter(x, y)
plt.show()
```

```r
import matplotlib.pyplot as plt
marks = [55, 60, 65, 70, 75, 80, 85, 90]
plt.boxplot(marks)
plt.show()
```

```r
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 170, 200]
plt.plot(months, sales, marker='o', color= 'blue')
plt.title("monthly sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.show()
```

```r
x <-c(1,2,3,4,5)
y <-c(10,15,12,18,20)
plot(x,
     y,
     type="o",
     col="blue",
     main="line graph",
     xlab="xvalues",
     ylab="yvalues")
```

```r
x <- c(1,2,3,4,5)
y <-c(10,15,12,18,20)
barplot(y,names.arg = x,col= "sky blue", main= "bar chart", xlab= "x values", ylab= "y values")
```

```r
x <-c (1,2,3,4,5)
y <-c (10,15,12,18,20)
plot (x,y,pch = 19,col ="red",main ="scatterplot",xlab ="x values",ylab ="y values")
```
